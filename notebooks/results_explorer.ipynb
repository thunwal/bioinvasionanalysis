## 2. Sensitivity analysis: which threshold?

The accumulated cost threshold removes least-cost paths that are too costly to be plausible annual dispersal. The
remaining connected paths form populations. The sensitivity test (`main.py` in `test` mode) repeats this for many
thresholds and records the average expansion rate of robust populations.

- **Steep increase:** fragments of the same population merge. The threshold is still below the dispersal capacity
  of the species.
- **Stable range:** the result hardly depends on the exact threshold. A good choice is at its beginning.
- **Jumps at the very top:** a few extreme paths merge separate introductions into implausible super-populations.

For a long tail of high thresholds, enter a value to split the x axis (the paper used 10).

```python
ex.sensitivity_section()
```

## 4. Expansion phases (optional)

The expansion rate is the slope of a linear model of the cumulative distance from the first observation over time.
If a population shows a lag phase or levels off at the end, a single model underestimates the expansion. Choose a
population, set the expansion phase (or click **Suggest phases**), and click **Use these phases** to include it in
the figures and in the phases table.

```python
ex.phases_section()
```
